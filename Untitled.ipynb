```python
import numpy as np
import pandas as pd
```

```python
order = pd.read_csv("order1.csv")
product = pd.read_csv("products_export_1.csv")
customer = pd.read_csv("customers_export.csv")
```

```python
product.head()
```

```python
order.head()
```

```python
order.dropna(axis=1, how='all', inplace=True)
product.dropna(axis=1, how='all', inplace=True)
customer.dropna(axis=1, how='all', inplace=True)
```

```python
order.drop_duplicates(inplace=True)

product.drop_duplicates(inplace=True)

customer.drop_duplicates(inplace=True)
```

```python
customer.isnull().sum()
```

```python
product.isnull().sum()
```

```python
order.isnull().sum()
```

```python
threshold = len(order) * 0.5

customer = customer.dropna(axis=1, thresh=threshold)
```

```python
threshold = len(order) * 0.5

order = order.dropna(axis=1, thresh=threshold)
```

```python
threshold = len(product) * 0.5

product = product.dropna(axis=1, thresh=threshold)
```

```python
text_columns = order.select_dtypes(include='object').columns

order[text_columns] = order[text_columns].fillna("null")
```

```python
text_columns = customer.select_dtypes(include='object').columns

customer[text_columns] = customer[text_columns].fillna("null")
```

```python
text_columns = product.select_dtypes(include='object').columns

product[text_columns] = product[text_columns].fillna("null")
```

```python
print(order['Created at'].dtype)
```

```python
order['Created at'] = pd.to_datetime(
    order['Created at'],
    errors='coerce'
)
```

```python
print(order['Created at'].dtype)
```

```python
order = order.copy()

order['Year'] = order['Created at'].dt.year
order['Month'] = order['Created at'].dt.month
order['Pdate'] = order['Created at'].dt.date
order['Day'] = order['Created at'].dt.day
order['hour'] = order['Created at'].dt.hour
```

```python
order["Email"].duplicated().sum()
```

```python
columns = ["Shipping Province Name", "Shipping Country", "Shipping City"]

order[columns] = order[columns].apply(lambda x: x.str.upper())
```
